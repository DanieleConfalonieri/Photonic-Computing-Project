```python
%matplotlib inline
import interferometer as inf
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt
```

Primo tentativo con il modulo interferometer di Clements

```python
U = np.array([[1/np.sqrt(2), 1/np.sqrt(2)], [1/np.sqrt(2), -1/np.sqrt(2)]])
Interf = inf.square_decomposition(U)
Interf.draw()
print(Interf.BS_list)
print(Interf.calculate_transformation()@np.array([[1],[0]]))
```

```python
V = inf.random_unitary(4)
Interf2 = inf.square_decomposition(V)
print(Interf2.BS_list)
print(Interf2.calculate_transformation())
for bs in Interf2.BS_list:
    bs.theta = bs.theta + 0.01 + np.imag(0.01) 
    bs.phi = bs.phi + 0.01 + np.imag(0.01)

print(Interf2.calculate_transformation())
```
